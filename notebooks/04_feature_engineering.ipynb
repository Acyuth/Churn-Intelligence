```python
# Generating the Feature Matrix
import pandas as pd
import sqlite3
from pathlib import Path
```

```python
# Connect to DB and run the feature query
conn = sqlite3.connect("../data/churn.db")
query = Path("../sql/03_features.sql").read_text()

# Load directly into Pandas
features_df = pd.read_sql(query, conn)
conn.close()

# Quick clean-up: Fill missing numeric features with 0 (since no logs = 0 listening)
features_df = features_df.fillna(0)

# Save the final ML-ready dataset to our processed folder
output_path = Path("../data/processed/features.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
features_df.to_csv(output_path, index=False)

print(f"Feature matrix saved to {output_path}!")
print(f"Final shape: {features_df.shape}")
print("\nPreview:")
print(features_df.head())
```
